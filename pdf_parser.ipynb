```python
# ============================================================
# 系统依赖说明（解析 PDF 时可能需要）
# ============================================================
# Python 包：pip install "unstructured[pdf]"
#
# Windows 额外系统依赖（按解析策略不同）：
# - strategy="fast"（默认较快）：一般只需 Python 包，不强制 poppler
# - strategy="hi_res"（高精度布局）：需要 poppler（pdf2image 用）
#     choco install poppler
#     或 https://github.com/oschwartz10612/poppler-windows/releases
# - strategy="ocr_only" 或 hi_res 扫描件：还需要 tesseract（OCR）
#     choco install tesseract
# ============================================================

from pathlib import Path

from unstructured.partition.pdf import partition_pdf

# PDF 路径（相对于 notebook 所在目录）
pdf_path = Path("workshops/其他格式/pdf/chevron-page.pdf")

# 解析 PDF，返回 Element 列表
# 初学者建议先用 fast 策略，避免 poppler/tesseract 未安装时报错
elements = partition_pdf(filename=str(pdf_path), strategy="fast")

# 统计 Element 总数
print(f"共解析出 {len(elements)} 个 Element\n")
print("前 10 个 Element：")
print("-" * 50)

for i, element in enumerate(elements[:10], start=1):
    # category 即元素类型，如 Title、NarrativeText、Table 等
    element_type = element.category
    # 文本过长时只显示前 120 个字符，避免输出卡顿
    text = element.text or ""
    preview = text if len(text) <= 120 else text[:120] + "..."

    print(f"[{i}] Type: {element_type}")
    print(f"    Text: {preview}\n")
```
